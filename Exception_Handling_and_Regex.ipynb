## Exercise 1: Building Fault-Tolerant Programs

### Part A: Division Calculator

```python
try:
    numerator = int(input("Enter a numerator: "))
    denominator = int(input("Enter a denominator: "))
    result = numerator / denominator
    print(f"The result is: {result}")
except ValueError:
    print("Error: Please enter valid integers.")
except ZeroDivisionError:
    print("Error: Denominator cannot be zero.")
finally:
    print("Program ended safely")
```

### Part B: Fault-Tolerant File Reader

```python
# Establishes a ValueError for an empty file.

with open("empty_file.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

    if not lines or all(not line.strip() for line in lines):
        raise ValueError("The file is empty!")

    for i, line in enumerate(lines[:5], start = 1):
        print(f"Line {i}: {line.strip()}")
```

```python
# Task without valueError for an empty file.

value = input("Enter a File Path:")

try:
    with open(value, "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("File not found.")
except PermissionError:
    print("Permission denied.")
except ValueError:
    print("File is empty.")

else:
    print("File read successfully.")
```

## Part C: Validating User Input

```python
def get_valid_age():

    while True:

        try:
            age = int(input('Enter your age:'))
            if type(age) != int:
                raise ValueError('Entered value not integer')
            if age < 1 or age > 120:
                raise ValueError('Age must be between 1 and 120')
            print(f'You are {age} years old!')
            break

        except ValueError as e:
            print('Error:',e,'\n Please try again')

    
```

## Exercise 2: Processing Text with Regular Expressions

### Part A: Extracting Contact Information

```python
text = """
Contact: Alice (alice.wonder@ncirl.ie), Bob (bob_smith@data.io)
Call us at +353-1-889-1234 or 01-7654321.
Visit https://www.ncirl.ie or http://learn.ncirl.ie for more info.
"""
```

```python
import re

emails = re.findall(r"[\w.-]+@[\w.-]+\.\w+",text)

irish_numbers = re.findall(r"(?:\+353|0)[\d-]+", text)

urls = re.findall(r"https?://[\w.-]+\.\w+", text)

print("Emails found:", emails)
print("Irish phone numbers found:", irish_numbers)
print("URLs found:", urls)
```

```python
contacts = {"emails": emails,
            "phones": irish_numbers,
            "urls": urls}

for category, values in contacts.items():
    print(f"{category}: {values}")
```

### Part B: Normalising Inconsistent Time Formats

```python
text = """
Morning session: 9am to 11:30 AM.
Lunch break: 12 PM - 1:15 pm.
Afternoon workshop: 14:0016:45.
Evening talk: 6.30 pm 8 pm.
"""
```

```python
pattern = r"\d{1,2}[:.]?\d{0,2}\s*(?:am\|pm)?"

times = re.findall(pattern, text, re.IGNORECASE)
print("Times found:", times)
```

```python
pattern = r"\d{1,2}[:.]?\d{0,2}\s*(?:am\|pm)?"

times = re.finditer(pattern, text, re.IGNORECASE)
for match in times:
    print("Time found:", match.group())
```

```python
pattern = r"(?\d{1,2})"r"[:.]?"r"(?\d{0,2})"r"\s*(?:am|pm)?"
print(pattern)
```

```python
def normalize_times(text):
    #pattern = r"\d{1,2}[:.]?\d{0,2}\s*(?:am\|pm)?"

    pattern = r"(?:\d{1,2})"r"[:.]?"r"(?:\d{0,2})"r"\s*(?:am|pm)?"
    

    matches = re.finditer(pattern, text, re.IGNORECASE) # gives different groups as outputs
    normalized_times = []
    for match in matches:
        hour = int(match.group(1))
        mins = int(match.group(2))
        period = match.group(3) # am or pm

        if mins > 59:
            continue  # Skip invalid minutes

        if period:
            period = period.lower()

            if hour <1 or hour > 12:
                continue  # Skip invalid hours
            if period == "pm" and hour != 12:
                hour += 12

            elif period == "am" and hour == 12:
                hour = 0
        
    formatted_time = f"{hour:02d}:{mins:02d}"
    normalized_times.append(formatted_time)
    return normalized_times

    
```

```python
def normalize_times_new(text):
    pattern = r"(\d{1,2})(?:[:.](\d{1,2}))?\s*(am|pm)?"

    matches = re.finditer(pattern, text, re.IGNORECASE)
    normalized_times = []

    for match in matches:
        hour = int(match.group(1))
        mins = int(match.group(2) or "0")
        period = match.group(3)

        if mins > 59:
            continue

        if period:
            period = period.lower()

            if hour < 1 or hour > 12:
                continue

            if period == "pm" and hour != 12:
                hour += 12
            elif period == "am" and hour == 12:
                hour = 0
        elif hour > 23:
            continue

        formatted_time = f"{hour:02d}:{mins:02d}"
        normalized_times.append(formatted_time)

    return normalized_times
```
